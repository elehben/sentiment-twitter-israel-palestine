### Cleaning data using wordsegment and nltk

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from langdetect import detect, LangDetectException
```

```python
import re #regex
from wordsegment import load, segment

load()

class TweetProcessor:
    def __init__(self, file_path):
        self.df = pd.read_csv(file_path)
        self.df['text'] = self.df['text'].astype(str)

    def clean_tweet(self, tweet):
        def fix_hashtags(text):
          hashtags = re.findall(r'#(\w+)', text)
          for ht in hashtags:
              broken_camel = re.sub(r'([a-z])([A-Z])', r'\1 \2', ht)

              segmented = segment(broken_camel)
              combined = ' '.join(segmented)

              text = text.replace(f'#{ht}', combined)
          return text

        tweet = fix_hashtags(tweet)

        tweet = re.sub(r'https\S+', '', tweet, flags=re.MULTILINE)

        tweet = re.sub(r'\W', ' ', tweet)

        tweet = re.sub(r'[^a-zA-Z0-9\s]', '', tweet)

        tweet = re.sub(r'\s+', ' ', tweet)

        tweet = tweet.strip()
        return tweet

    def tokenize_and_lemmatize(self, tweet):
        tokens = word_tokenize(tweet)
        tokens = [word.lower() for word in tokens if word.isalpha()]
        stop_words = set(stopwords.words('english'))
        tokens = [word for word in tokens if word not in stop_words]
        lemmatizer = WordNetLemmatizer()
        tokens = [lemmatizer.lemmatize(word) for word in tokens]
        return ' '.join(tokens)

    def drop_unused_columns(self):
        columns_to_drop = ['id', 'link', 'date', 'likes', 'comments']
        self.df = self.df.drop(columns=[col for col in columns_to_drop if col in self.df.columns])

    def process_tweets(self):
        def lang(x):
            try:
                if not isinstance(x, str) or not x.strip():
                    return False
                return detect(x) == 'en'
            except LangDetectException:
                return False

        self.df = self.df[self.df['text'].apply(lang)]
        self.df['cleaned_text'] = self.df['text'].apply(self.clean_tweet)
        self.df['tokenized_and_lemmatized'] = self.df['cleaned_text'].apply(self.tokenize_and_lemmatize)
        self.drop_unused_columns()
```

### Labelling Data using vaderSentiment

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

new_words = {
    'genocide': -4.0,
    'hostage': -3.5,
    'war': -2.0,
    'attack': -2.0,
    'slaughtered': -4.0,
    'atrocity': -4.0,
    'terrorism': -4.0
}
analyzer.lexicon.update(new_words)

def get_sentiment_label(text):
    if not isinstance(text, str) or text.strip() == '':
        return 'Neutral'
    score = analyzer.polarity_scores(text)
    compound = score['compound']

    if compound >= 0.04:
        return 'Positive'
    elif compound <= -0.04:
        return 'Negative'
    else:
        return 'Neutral'

processor.df['sentiment'] = processor.df['tokenized_and_lemmatized'].apply(get_sentiment_label)

print("Distribusi Sentimen:")
display(processor.df['sentiment'].value_counts())

display(processor.df[['tokenized_and_lemmatized', 'sentiment']].head(10))
```

### Splitting Data and Encoding using TF-IDF

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.corpus import stopwords

custom_stopwords = ['israel', 'israeli', 'hamas', 'gaza', 'palestine', 'palestinian', 'people', 'said', 'go']
final_stopwords = list(set(stopwords.words('english')).union(set(custom_stopwords)))

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    processor.df_balanced['tokenized_and_lemmatized'].astype(str),
    processor.df_balanced['sentiment'],
    test_size=0.2,
    random_state=42,
    stratify=processor.df_balanced['sentiment'])

tfidf = TfidfVectorizer(max_features=5000, stop_words=final_stopwords)

X_train = tfidf.fit_transform(X_train_raw)

X_test = tfidf.transform(X_test_raw)

print(f"Data Latih: {X_train.shape}")
print(f"Data Uji : {X_test.shape}")
```

### Training and Testing Model using Naive Bayes, SVM, and BERT

### Training Model

```python
import torch
from torch.utils.data import DataLoader, TensorDataset
from transformers import BertTokenizer, BertForSequenceClassification
from torch.optim import AdamW
from sklearn.preprocessing import LabelEncoder
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup
from sklearn.model_selection import GridSearchCV
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.metrics import classification_report

# --- Naive Bayes ---
param_grid_nb = {'alpha': [0.1, 0.5, 1.0, 2.0, 5.0]}
grid_nb = GridSearchCV(MultinomialNB(), param_grid_nb, cv=5, scoring='accuracy', n_jobs=-1)
grid_nb.fit(X_train, y_train)

nb_results = pd.DataFrame({
    'Alpha'         : grid_nb.cv_results_['param_alpha'],
    'Mean CV Acc'   : grid_nb.cv_results_['mean_test_score'],
    'Std CV Acc'    : grid_nb.cv_results_['std_test_score'],
    'Rank'          : grid_nb.cv_results_['rank_test_score']
}).sort_values('Mean CV Acc', ascending=False)

nb_results['Mean CV Acc'] = nb_results['Mean CV Acc'].map('{:.4f}'.format)
nb_results['Std CV Acc']  = nb_results['Std CV Acc'].map('± {:.4f}'.format)

print("=== Rata-rata Akurasi CV per Alpha — Naive Bayes ===")
print(nb_results.to_string(index=False))

# --- SVM ---
param_grid_svm = {
    'C': [0.1, 1, 10],
    'kernel': ['linear', 'rbf']
}
grid_svm = GridSearchCV(SVC(random_state=42), param_grid_svm, cv=5, scoring='accuracy', n_jobs=-1)
grid_svm.fit(X_train, y_train)

svm_results = pd.DataFrame({
    'C'           : grid_svm.cv_results_['param_C'],
    'Kernel'      : grid_svm.cv_results_['param_kernel'],
    'Mean CV Acc' : grid_svm.cv_results_['mean_test_score'],
    'Std CV Acc'  : grid_svm.cv_results_['std_test_score'],
    'Rank'        : grid_svm.cv_results_['rank_test_score']
}).sort_values('Mean CV Acc', ascending=False)

svm_results['Mean CV Acc'] = svm_results['Mean CV Acc'].map('{:.4f}'.format)
svm_results['Std CV Acc']  = svm_results['Std CV Acc'].map('± {:.4f}'.format)

print("\n=== Rata-rata Akurasi CV per Kombinasi Parameter — SVM ===")
print(svm_results.to_string(index=False))

# --- BERT ---
le = LabelEncoder()
y_train_le = le.fit_transform(y_train)
y_test_le = le.transform(y_test)

model_name = "cardiffnlp/twitter-roberta-base-sentiment-latest"

# Load Tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

def encode_data(texts, labels):
    encodings = tokenizer(list(texts), truncation=True, padding=True, max_length=128, return_tensors='pt')
    dataset = TensorDataset(encodings['input_ids'], encodings['attention_mask'], torch.tensor(labels))
    return dataset

train_dataset = encode_data(X_train_raw, y_train_le)
test_dataset = encode_data(X_test_raw, y_test_le)

train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=16)

# Load BERT Model
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model_bert = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=3).to(device)

optimizer = AdamW(model_bert.parameters(), lr=2e-5, weight_decay=0.01)

epochs = 3
total_steps = len(train_loader) * epochs

scheduler = get_linear_schedule_with_warmup(
    optimizer,
    num_warmup_steps=int(0.1 * total_steps),
    num_training_steps=total_steps
)

train_losses, val_losses         = [], []
train_accuracies, val_accuracies = [], []

best_val_loss = float('inf')
patience = 2
patience_counter = 0

# Training loop
print(f"\nTraining BERT on {device}...")
for epoch in range(epochs):
    print(f"======== Epoch {epoch+1} / {epochs} ========")
    model_bert.train()
    total_train_loss = 0
    train_correct       = 0
    train_total         = 0
    for batch in train_loader:
        optimizer.zero_grad()
        input_ids, attention_mask, labels = [b.to(device) for b in batch]

        outputs = model_bert(input_ids, attention_mask=attention_mask, labels=labels)
        loss = outputs.loss
        total_train_loss += loss.item()

        preds          = torch.argmax(outputs.logits, dim=-1)
        train_correct += (preds == labels).sum().item()
        train_total   += labels.size(0)

        loss.backward()
        optimizer.step()
        scheduler.step()

    avg_train_loss = total_train_loss / len(train_loader)
    avg_train_acc  = train_correct / train_total

    print(f"  Training Loss   : {avg_train_loss:.4f}")
    print(f"  Training Accuracy : {avg_train_acc:.4f}")

    model_bert.eval()
    total_val_loss = 0
    val_correct    = 0
    val_total      = 0

    with torch.no_grad():
        for batch in test_loader:
            input_ids, attention_mask, labels = [b.to(device) for b in batch]
            outputs = model_bert(input_ids, attention_mask=attention_mask, labels=labels)
            loss = outputs.loss
            total_val_loss += loss.item()

            preds        = torch.argmax(outputs.logits, dim=-1)
            val_correct += (preds == labels).sum().item()
            val_total   += labels.size(0)

    avg_val_loss = total_val_loss / len(test_loader)
    avg_val_acc  = val_correct / val_total

    print(f"  Validation Loss : {avg_val_loss:.4f}")
    print(f"  Validation Acc    : {avg_val_acc:.4f}")

    train_losses.append(avg_train_loss)
    val_losses.append(avg_val_loss)
    train_accuracies.append(avg_train_acc)
    val_accuracies.append(avg_val_acc)

    # Logika Early Stopping
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        patience_counter = 0
        torch.save(model_bert.state_dict(), 'best_model_bert.pt')
        print("  => Model membaik! Disimpan ke best_model_bert.pt")
    else:
        patience_counter += 1
        print(f"  => Tidak ada perbaikan pada Validation Loss (Patience: {patience_counter}/{patience})")
        if patience_counter >= patience:
            print("  => Early Stopping Terpicu! Proses Training Dihentikan.")
            break

print("\n=== Rata-rata Akurasi Training BERT per Epoch ===\n")

bert_history = pd.DataFrame({
    'Epoch'         : list(range(1, len(train_losses) + 1)),
    'Train Loss'    : [f"{v:.4f}" for v in train_losses],
    'Val Loss'      : [f"{v:.4f}" for v in val_losses],
    'Train Acc (%)'  : [f"{v*100:.2f}%" for v in train_accuracies],
    'Val Acc (%)'    : [f"{v*100:.2f}%" for v in val_accuracies],
})

print(bert_history.to_string(index=False))

best_epoch = val_losses.index(min(val_losses)) + 1
print(f"\nBest epoch berdasarkan val loss terendah : Epoch {best_epoch}")
print(f"Train Acc di epoch terbaik               : {train_accuracies[best_epoch-1]*100:.2f}%")
print(f"Val Acc di epoch terbaik                 : {val_accuracies[best_epoch-1]*100:.2f}%")
```

```python
print("=== Hasil Hyperparameter Tuning ===\n")

print("1. Naive Bayes (MultinomialNB)")
print(f"   - Best Alpha: {grid_nb.best_params_['alpha']}")
print(f"   - Best Accuracy Score: {grid_nb.best_score_:.4f}")

print("\n2. Support Vector Machine (SVC)")
print(f"   - Best C: {grid_svm.best_params_['C']}")
print(f"   - Best Kernel: {grid_svm.best_params_['kernel']}")
print(f"   - Best Accuracy Score: {grid_svm.best_score_:.4f}")

print("\n3. BERT (RoBERTa - Manual Tuning)")
print(f"   - Learning Rate: 2e-5")
print(f"   - Weight Decay: 0.01")
print(f"   - Epochs: {epochs}")
print(f"   - Scheduler: Linear with Warmup (10% steps)")
print(f"   - Best Validation Loss Achieved: {best_val_loss:.4f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

labels = ['Negative', 'Neutral', 'Positive']

def plot_confusion_matrix(y_true, y_pred, title, ax):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    sns.heatmap(cm, annot=True, fmt='d', cmap='RdPu',
                xticklabels=labels, yticklabels=labels, ax=ax)
    ax.set_title(title)
    ax.set_ylabel('Actual (Kenyataan)')
    ax.set_xlabel('Predicted (Prediksi)')

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(18, 5))

# Plot untuk Naive Bayes
plot_confusion_matrix(y_test, y_pred_nb, 'Confusion Matrix: Naive Bayes', ax1)

# Plot untuk SVM
plot_confusion_matrix(y_test, y_pred_svm, 'Confusion Matrix: SVM', ax2)

# Plot untuk BERT
plot_confusion_matrix(y_test, y_pred_bert_labels, 'Confusion Matrix: BERT', ax3)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

cm_svm = confusion_matrix(y_test, y_pred_svm, labels=['Negative', 'Neutral', 'Positive'])
cm_nb = confusion_matrix(y_test, y_pred_nb, labels=['Negative', 'Neutral', 'Positive'])

acc_svm = cm_svm.diagonal() / cm_svm.sum(axis=1)
acc_nb = cm_nb.diagonal() / cm_nb.sum(axis=1)
acc_bert_total = accuracy_score(y_test, y_pred_bert_labels)

class_names = ['Negative', 'Neutral', 'Positive']
cm_bert = confusion_matrix(y_test, y_pred_bert_labels, labels=class_names)
acc_per_class_bert = cm_bert.diagonal() / cm_bert.sum(axis=1)

models = ['SVM', 'Naive Bayes', 'BERT']
all_accuracies = []
all_labels = []
all_model_names = []

for model_name, acc_values in zip(models, [acc_svm, acc_nb, acc_per_class_bert]):
    for i, label in enumerate(class_names):
        all_accuracies.append(acc_values[i])
        all_labels.append(label)
        all_model_names.append(model_name)

data_acc_final = pd.DataFrame({
    'Sentiment': all_labels,
    'Accuracy': all_accuracies,
    'Model': all_model_names
})

plt.figure(figsize=(12, 7))
sns.barplot(data=data_acc_final, x='Sentiment', y='Accuracy', hue='Model', palette='magma')
plt.title('Perbandingan Akhir Akurasi: SVM vs Naive Bayes vs BERT')
plt.ylim(0, 1.1)
plt.ylabel('Akurasi (Score)')
plt.grid(axis='y', linestyle='--', alpha=0.6)

for p in plt.gca().patches:
    if p.get_height() > 0:
        plt.gca().annotate(f'{p.get_height():.2%}',
                       (p.get_x() + p.get_width() / 2., p.get_height()),
                       ha='center', va='center', xytext=(0, 9),
                       textcoords='offset points')

plt.show()
```
